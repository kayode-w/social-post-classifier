```python
import requests
import json
from datetime import datetime, timedelta, timezone

url = f"https://mastodon.social/api/v1/timelines/tag/tech"

response = requests.get(url, params={"limit": 40}, timeout=10) #limit to 40 posts and timeout after 10 seconds
response.raise_for_status()  #raise an error for bad responses
data = response.json()

posts = data
```

```python
print(list(posts[0].keys()))
```

```python
INSTANCE = 'mastodon.social'
HASHTAG = 'fitness'
current_time = datetime.now(timezone.utc).isoformat()

tbl = []

row = {
    "post_id": posts[1]["id"],
    "content": posts[1]["content"],
    "created_at": posts[1]["created_at"],
    "language": posts[1]["language"],
    "is_reblog": False if posts[1]["reblog"] is None else True,
    "instance": INSTANCE,
    "hashtag": HASHTAG,
    "loaded_at": current_time
}

tbl.append(row)

print(json.dumps(row, indent=2))
```
